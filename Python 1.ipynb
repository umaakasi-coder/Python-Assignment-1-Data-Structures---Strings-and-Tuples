```python
# String Concatenation Example

string1 = "Hello "
string2 = "Umaa "   # replace with your name
string3 = "Welcome to Python programming"
result = string1 + string2 + string3

print(result)

# 2. Print the first character of the string

print("First character:", result[0])

# Print the last character of the string.

print("Last character:", result[-1])

#Print the first 5 characters of the string.

print("First 5 characters:", result[:5])

#Print the last 11 characters of the string.

print("Last 11 characters:", result[-11:])

#Print the string in reverse.

print("Reversed string:", result[::-1])

# Use slicing to extract "Python"

print("Extracted word:", result[22:29])







   


```
