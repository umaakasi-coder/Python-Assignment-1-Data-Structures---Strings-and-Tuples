```python
 # 3.Define the string

strM = "Python beginner tutorial"

# Convert to uppercase

upper_str = strM.upper()

# Print the result

print("Uppercase sentence:", upper_str)

# Convert to lowercase

lower_str = strM.lower()

# Print the result
print("Lowercase sentence:", lower_str)

# Convert using capitalize
cap_str = strM.capitalize()

# Print the result
print("Capitalized sentence:", cap_str)

# Count occurrences of 't'
count_t = strM.count('t')

# Print the result
print("Total occurrences of 't':", count_t)

# Replace "Python" with "Machine Learning"
new_str = strM.replace("Python", "Machine Learning")

# Print the result
print("Modified sentence:", new_str)

```

```python
# Create the tuples

tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# Print both tuples
print("First tuple:", tuple1)
print("Second tuple:", tuple2)

```

```python
# Concatenate the tuples
t_combine = tuple1 + tuple2

# Print the result
print("Combined tuple:", t_combine)
```

```python
# Repeat the elements of t_combine 3 times
t_repeat = t_combine * 3

# Print the result
print("Repeated tuple:", t_repeat)

```

```python
# Access the 3rd element (index 2)
third_element = t_combine[2]

# Print the result
print("3rd element from t_combine:", third_element)
```

```python
# Access the first three elements
first_three = t_combine[:3]

# Print the result
print("First three elements from t_combine:", first_three)
```

```python
# Access the last three elements
last_three = t_combine[-3:]

# Print the result
print("Last three elements from t_combine:", last_three)
```
